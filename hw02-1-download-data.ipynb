## Downloading Hourly Wikipedia Event Data

Importing libraries

```python
import boto3
from pathlib import Path
```

Accessing data from S3 bucket

```python
# setting up S3 call and data directory
s3 = boto3.client("s3")
data_dir = Path("data")
data_dir.mkdir(exist_ok = True)

# using https://docs.aws.amazon.com/code-library/latest/ug/python_3_s3_code_examples.html for inspo
response = s3.list_objects_v2(Bucket = "dsan6000-wikipedia", Prefix = "hourly_parquet/")
```

Saving all .parquet files

```python
for item in response.get("Contents"):
    key = item["Key"]

    if key.endswith(".parquet"):
        filename = Path(key).name
        s3.download_file("dsan6000-wikipedia", key, str(data_dir / filename))    
```
