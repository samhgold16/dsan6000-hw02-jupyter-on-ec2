# Moving Wikipedia Data into S3 Bucket

Importing libraries

```python
import boto3
from pathlib import Path
```

Creating bucket and moving files into it

```python
# setting up data and S3 initializations
s3 = boto3.client("s3")
data_dir = Path("data")
data_dir.mkdir(exist_ok = True)

# bucket name
bucket_name = "dsan6000-sg1951"

# creating bucket
bucket = s3.create_bucket(Bucket = bucket_name)

# moving data
for file in data_dir.iterdir():
    s3.upload_file(str(file), bucket_name, f"wikipedia-hourly/{file.name}")
```
